## 1️⃣ What is Feature Selection?
Feature Selection is the process of selecting the most important and relevant features (columns) from the dataset.  
It helps in:
- Reducing overfitting  
- Improving model accuracy  
- Faster computation  
- Better interpretability  

---


```python
# Import Libraries
import pandas as pd
import numpy as np
from sklearn.datasets import load_wine
from sklearn.feature_selection import SelectKBest, f_classif

# Load dataset
data = load_wine()
X, y = data.data, data.target
df = pd.DataFrame(X, columns=data.feature_names)

print("Original Dataset Shape:", X.shape)
df.head()
```

## 2️⃣ Feature Selection Example (SelectKBest)
We will use `SelectKBest` with the `f_classif` score function to pick the **top 5 features**.


```python
# Feature Selection
selector = SelectKBest(score_func=f_classif, k=5)
X_new = selector.fit_transform(X, y)

selected_features = np.array(data.feature_names)[selector.get_support()]
print("Selected Top 5 Features:", selected_features)
```

## 3️⃣ Feature Extraction
Feature Extraction transforms the original features into a new feature space.  
Example: **Principal Component Analysis (PCA)** reduces dimensionality while keeping maximum variance.


```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA (2D projection)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Explained Variance Ratio:", pca.explained_variance_ratio_)
```

## 4️⃣ Curse of Dimensionality
The **curse of dimensionality** refers to problems that arise when data has too many features:
- Distance metrics (like in KNN) become less meaningful.  
- Models may overfit with too many features.  
- Computational cost increases drastically.  

We will compare **KNN accuracy** on the full dataset vs PCA-reduced dataset.


```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

# KNN on full features
knn_full = KNeighborsClassifier(n_neighbors=5)
knn_full.fit(X_train, y_train)
y_pred_full = knn_full.predict(X_test)
print("KNN Accuracy with all features:", accuracy_score(y_test, y_pred_full))

# KNN on PCA-reduced features
X_train_pca, X_test_pca, y_train_pca, y_test_pca = train_test_split(X_pca, y, test_size=0.3, random_state=42)
knn_pca = KNeighborsClassifier(n_neighbors=5)
knn_pca.fit(X_train_pca, y_train_pca)
y_pred_pca = knn_pca.predict(X_test_pca)
print("KNN Accuracy with PCA reduced features:", accuracy_score(y_test_pca, y_pred_pca))
```

## ✅ Summary
- **Feature Selection**: Helps in choosing the best subset of features.  
- **Feature Extraction (PCA)**: Creates new transformed features for dimensionality reduction.  
- **Curse of Dimensionality**: Too many features can harm model performance.
