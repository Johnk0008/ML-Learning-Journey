# Feature Construction & Feature Splitting

## Import Libraries

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import cross_val_score 
from sklearn.linear_model import LogisticRegression 

import seaborn as sns
```

## Import Dataset

```python
 df = pd.read_csv('titanic.csv')
```

```python
df.head()
```

```python
df = pd.read_csv('titanic.csv')[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

# Drop NA Value

```python
df.dropna(inplace=True)
```

```python
df.sample(5)
```

# Separate X and Y

```python
X = df.iloc[:,0:4] 
y = df.iloc[:,-1]
```

```python
X.head()
```

# Check Accuracy for Logistic Regression

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20)) * 100
```

# Applying Feature Construction

## Create New Column

```python
X['Family_size'] = X['SibSp'] + X['Parch'] + 1
```

```python
X.head()
```

## Apply New Function

```python
def myfunc(num): 
    if num == 1:
      #alone
      return 0
    elif num >1 and num <=4:
      # small family
      return 1 
    else:
      # large family
      return 2
```

```python
myfunc(4)
```

## Apply M Function

```python
X['Family_type'] = X['Family_size'].apply(myfunc)
```

```python
X.head()
```

# Drop unwanted columns

X.drop(columns=['SibSp','Parch','Family_size'],inplace=True)

X.head()

# Review Accuracy after Feature Construction

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20)) *100
```
